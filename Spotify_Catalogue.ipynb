# ***SPOTIFY CATALOGUE***
**DESCRIPTION OF PROJECT :**

This project analyses a Spotify tracks dataset from Kaggle using Python, pandas and numpy. It loads the data, explores it, and cleans it by removing missing values, duplicates and zero-popularity songs. It then answers simple questions, such as which songs, artists and genres are the most popular, and whether explicit or short songs do better. It also checks how audio features like danceability, energy and tempo relate to popularity. Finally, numpy is used to label songs by mood and popularity level, calculate tempo statistics, and save the results into two CSV files.

```python
import numpy as np
import pandas as pd


df = pd.read_csv("dataset.csv")
print("Shape:", df.shape)
print(df.head())
df.info()


print(df.describe().round(2))
print(df.isnull().sum())


before = len(df)
df = df.dropna()
df = df.drop_duplicates(subset="track_id")
df = df[df["popularity"] > 0]
df["duration_min"] = np.round(df["duration_ms"] / 60000, 2)
print("Rows before:", before, "| after:", len(df))


top10 = df.sort_values("popularity", ascending=False).head(10)
print(top10[["track_name", "artists", "popularity"]])

print(df["artists"].value_counts().head(5))


genre_pop = df.groupby("track_genre")["popularity"].mean().round(2)
genre_pop = genre_pop.sort_values(ascending=False)
print(genre_pop)


print(df.groupby("explicit")["popularity"].mean().round(2))


df["length"] = np.where(df["duration_min"] < 3.5, "Short", "Long")
print(df.groupby("length")["popularity"].agg(["mean", "count"]).round(2))


features = ["danceability", "energy", "valence", "tempo", "duration_min"]
corr = df[features + ["popularity"]].corr()["popularity"].drop("popularity")
print(corr.sort_values(ascending=False).round(3))

df["mood"] = np.where(df["valence"] >= 0.5, "Happy", "Sad/Calm")
df["level"] = np.select([df["popularity"] >= 70, df["popularity"] >= 40],
                        ["Hit", "Average"], default="Low")
df["party_song"] = np.where((df["danceability"] > 0.7) & (df["energy"] > 0.7), "Yes", "No")
print(df["mood"].value_counts())
print(df["level"].value_counts())
print(df.groupby("party_song")["popularity"].mean().round(2))


tempo = df["tempo"].to_numpy()
print("Mean:", round(np.mean(tempo), 2))
print("Median:", round(np.median(tempo), 2))
print("Std:", round(np.std(tempo), 2))
print("90th percentile:", round(np.percentile(tempo, 90), 2))


genre_pop.to_csv("genre_popularity.csv")
df[df["level"] == "Hit"].to_csv("hit_songs.csv", index=False)
print("Saved genre_popularity.csv and hit_songs.csv")
```
